```python
    import pandas as pd

    df = pd.read_csv(
        "student_placement_prediction_feature_engineered.csv"
    )

    print(df.shape)
    df.head()

    def recommend_role(row):

        roles = []

        if (
            row["coding_skill_score"] >= 0.75 and
            row["projects_count"] >= 0.384615 and
            row["github_repos"] >= 0.30 and
            row["internships_count"] >= 0.15
        ):
            roles.append("Software Developer")

        if (
            row["coding_skill_score"] >= 0.65 and
            row["cgpa"] >= 7.5 and
            row["projects_count"] >= 0.230769
        ):
            roles.append("Backend Developer")

        if (
            row["coding_skill_score"] >= 0.60 and
            row["github_repos"] >= 0.20 and 
            row["projects_count"] >= 0.230769
        ):
            roles.append("Frontend Developer")

        if (
            row["aptitude_score"] >= 0.70 and
            row["logical_reasoning_score"] >= 0.70
        ):
            roles.append("Data Analyst")

        if (
            row["communication_skill_score"] >= 0.72 and
            row["leadership_score"] >= 0.65
        ):
            roles.append("Business Analyst")

        if (
            row["communication_skill_score"] >= 0.75 and
            row["mock_interview_score"] >= 0.70
        ):
            roles.append("HR / Talent Acquisition")

        if (
            row["coding_skill_score"] >= 0.60 and
            row["logical_reasoning_score"] >= 0.65 and 
            row["aptitude_score"] >= 0.60
        ):
            roles.append("QA/Test Engineer")

        if (
            row["coding_skill_score"] >= 0.70 and
            row["github_repos"] >= 0.30 and
            row["internships_count"] >= 0.15 and 
            row["logical_reasoning_score"] >= 0.60
        ):
            roles.append("DevOps Engineer")

        if (
            row["coding_skill_score"] >= 0.75 and
            row["projects_count"] >= 0.307692 and
            row["github_repos"] >= 0.25
        ):
            roles.append("Full Stack Developer")

        if (
            row["aptitude_score"] >= 0.75 and
            row["logical_reasoning_score"] >= 0.75 and
            row["coding_skill_score"] >= 0.65
        ):
            roles.append("Data Scientist")

        if (
            row["logical_reasoning_score"] >= 0.75 and
            row["coding_skill_score"] >= 0.60 and
            row["github_repos"] >= 0.20
        ):
            roles.append("Cybersecurity Analyst")

        if (
            row["coding_skill_score"] >= 0.65 and
            row["github_repos"] >= 0.25 and
            row["internships_count"] >= 0.15
        ):
            roles.append("Cloud Engineer")

        if len(roles) == 0:
            roles.append("IT Associate")

        return ", ".join(roles)
        
    df["recommended_role"] = df.apply(
        recommend_role,
        axis=1
    )

    print(
        df["recommended_role"]
        .value_counts()
    )

    def skill_gap(row):

        gaps = []

        if row["coding_skill_score"] < 0.60:
            gaps.append("Coding Skills")

        if row["communication_skill_score"] < 0.60:
            gaps.append("Communication Skills")
        
        if row["logical_reasoning_score"] < 0.60:
            gaps.append("Logical Reasoning")
        
        if row["aptitude_score"] < 0.60:
            gaps.append("Aptitude Skills")
        
        if row["projects_count"] < 0.230769:      # <3 projects
            gaps.append("Projects")
        
        if row["internships_count"] < 0.153846:   # <2 internships
            gaps.append("Internships")
        
        if row["github_repos"] < 0.20:
            gaps.append("GitHub Portfolio")
        
        if row["mock_interview_score"] < 0.60:
            gaps.append("Interview Preparation")
        
        if row["leadership_score"] < 0.50:
            gaps.append("Leadership Skills")
        
        if row["resume_score"] < 0.60:
            gaps.append("Resume Quality")
        
        if row["linkedin_connections"] < 0.20:
            gaps.append("Professional Networking")

        if len(gaps) == 0:
            gaps.append("No major skill gaps identified")

        return ", ".join(gaps)

    df["skill_gap"] = df.apply(
        skill_gap,
        axis=1
    )

    def recommendations(row):

        suggestions = []

        if row["coding_skill_score"] < 0.60:
            suggestions.append(
                "Improve DSA and coding practice"
            )

        if row["logical_reasoning_score"] < 0.60:
            suggestions.append(
                "Practice aptitude and logical reasoning questions"
            )

        if row["projects_count"] < 0.230769:
            suggestions.append(
                "Build more real-world projects"
            )

        if row["internships_count"] < 0.153846:
            suggestions.append(
                "Complete at least one internship"
            )

        if row["github_repos"] < 0.20:
            suggestions.append(
                "Maintain an active GitHub portfolio"
            )

        if row["communication_skill_score"] < 0.60:
            suggestions.append(
                "Improve communication and presentation skills"
            )

        if row["mock_interview_score"] < 0.60:
            suggestions.append(
                "Participate in mock interviews regularly"
            )

        if row["leadership_score"] < 0.50:
            suggestions.append(
                "Take part in leadership and team activities"
            )

        if row["resume_score"] < 0.60:
            suggestions.append(
                "Improve resume quality and ATS optimization"
            )

        if row["linkedin_connections"] < 0.20:
            suggestions.append(
                "Build a stronger professional network on LinkedIn"
            )

        if row["aptitude_score"] < 0.60:
            suggestions.append(
                "Practice aptitude questions regularly"
            )

        if len(suggestions) == 0:
            suggestions.append(
                "You are placement ready. Continue improving your portfolio and interview preparation."
            )
        return ", ".join(suggestions)

    df["recommendations"] = df.apply(
        recommendations,
        axis=1
    )

    print(
        df[
            [
                "recommended_role",
                "skill_gap",
                "recommendations"
            ]
        ].head(10)
    )

    import matplotlib.pyplot as plt

    plt.figure(figsize=(10,6))

    from collections import Counter

    all_roles = []

    for roles in df["recommended_role"]:
        role_list = roles.split(", ")
        all_roles.extend(role_list)

    role_counts = pd.Series(
        Counter(all_roles)
    ).sort_values(
        ascending=False
    )

    plt.figure(figsize=(12,6))

    role_counts.plot(kind="bar")

    plt.title(
        "Career Recommendation Distribution"
    )

    plt.xlabel("Job Role")
    plt.ylabel("Recommendations")

    plt.xticks(rotation=45)

    plt.tight_layout()

    plt.show()

    print(df["projects_count"].describe())
    print(df["projects_count"].value_counts().head(10))
    print(df["coding_skill_score"].describe())
```
