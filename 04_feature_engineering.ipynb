```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt

df = pd.read_csv("student_placement_prediction_dataset_2026.csv")

score_features = [
    "internships_count",
    "projects_count",
    "certifications_count",
    "hackathons_participated",
    "github_repos",
    "linkedin_connections",
    "coding_skill_score",
    "aptitude_score",
    "communication_skill_score",
    "logical_reasoning_score",
    "mock_interview_score",
    "leadership_score"
]

scaler = MinMaxScaler()

df[score_features] = scaler.fit_transform(df[score_features])

df["resume_score"] = (
    0.25 * df["projects_count"] +
    0.20 * df["internships_count"] +
    0.15 * df["certifications_count"] +
    0.15 * df["github_repos"] +
    0.10 * df["hackathons_participated"] +
    0.10 * df["leadership_score"] +
    0.05 * df["linkedin_connections"]
) * 100

df["employability_score"] = (
    0.25 * df["coding_skill_score"] +
    0.20 * df["aptitude_score"] +
    0.20 * df["communication_skill_score"] +
    0.15 * df["logical_reasoning_score"] +
    0.10 * df["mock_interview_score"] +
    0.10 * df["leadership_score"]
) * 100

df["recruitment_score"] = (
    0.25 * df["coding_skill_score"] +
    0.20 * df["aptitude_score"] +
    0.20 * df["communication_skill_score"] +
    0.15 * df["projects_count"] +
    0.10 * df["internships_count"] +
    0.10 * df["leadership_score"]
) * 100

print("\n===== OVERALL STATISTICS =====")

print(
    df[
        [
            "resume_score",
            "employability_score",
            "recruitment_score"
        ]
    ].describe().round(2)
)

scores = [
    "resume_score",
    "employability_score",
    "recruitment_score"
]

for score in scores:
    plt.figure(figsize=(8,5))
    plt.hist(df[score], bins=30)

    plt.title(f"{score.replace('_', ' ').title()} Distribution")
    plt.xlabel("Score")
    plt.ylabel("Number of Students")

    plt.show()

score_columns = [
    "resume_score",
    "employability_score",
    "recruitment_score"
]

tier_scores = (
    df.groupby("college_tier")[score_columns]
    .describe()
)

print("\n===== TIER-WISE SCORE ANALYSIS =====")
print(tier_scores.round(2))

branch_scores = (
    df.groupby("branch")[score_columns]
    .describe()
)

print("\n===== BRANCH-WISE SCORE ANALYSIS =====")
print(branch_scores.round(2))

tier_mean_scores = (
    df.groupby("college_tier")[score_columns]
    .mean()
)

tier_mean_scores.plot(
    kind="bar",
    figsize=(10,6)
)

plt.title("Tier-wise Average Scores")
plt.xlabel("College Tier")
plt.ylabel("Score")
plt.xticks(rotation=0)
plt.legend()
plt.show()

branch_mean_scores = (
    df.groupby("branch")[score_columns]
    .mean()
)

branch_mean_scores.plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("Branch-wise Average Scores")
plt.xlabel("Branch")
plt.ylabel("Score")
plt.xticks(rotation=0)
plt.legend()
plt.show()

df.to_csv(
    "student_placement_prediction_feature_engineered.csv",
    index=False
)
```
