```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("student_placement_prediction_dataset_2026.csv")

placed = df[df["placement_status"] == "Placed"]
not_placed = df[df["placement_status"] == "Not Placed"]

features = [
    "cgpa",
    "coding_skill_score",
    "aptitude_score",
    "communication_skill_score",
    "logical_reasoning_score",
    "mock_interview_score",
    "internships_count",
    "projects_count",
    "certifications_count",
    "backlogs"
]

for tier in df["college_tier"].unique():

    print("\n" + "="*80)
    print(f"{tier} ANALYSIS")
    print("="*80)

    tier_df = df[df["college_tier"] == tier]

    placed = tier_df[tier_df["placement_status"] == "Placed"]
    not_placed = tier_df[tier_df["placement_status"] == "Not Placed"]

    placed_means = []
    not_placed_means = []

    for feature in features:
        print(f"\n{feature}")

        print(
            f"Placed     -> Mean: {placed[feature].mean():.2f}"
            f" | Median: {placed[feature].median():.2f}"
        )

        print(
            f"Not Placed -> Mean: {not_placed[feature].mean():.2f}"
            f" | Median: {not_placed[feature].median():.2f}"
        )

    for feature in features:
        placed_means.append(placed[feature].mean())
        not_placed_means.append(not_placed[feature].mean())

    comparison_df = pd.DataFrame({
        "Placed Mean": placed_means,
        "Not Placed Mean": not_placed_means
    }, index=features)

    comparison_df.plot(
        kind="bar",
        figsize=(14,6)
    )

    plt.title(f"{tier} - Placed vs Not Placed Feature Comparison")
    plt.xlabel("Features")
    plt.ylabel("Average Value")
    plt.xticks(rotation=45)
    plt.legend()
    plt.show()

for branch in df["branch"].unique():

    print("\n" + "="*80)
    print(f"{branch} ANALYSIS")
    print("="*80)

    branch_df = df[df["branch"] == branch]

    placed = branch_df[branch_df["placement_status"] == "Placed"]
    not_placed = branch_df[branch_df["placement_status"] == "Not Placed"]

    placed_means = []
    not_placed_means = []

    for feature in features:
        print(f"\n{feature}")

        print(
            f"Placed     -> Mean: {placed[feature].mean():.2f}"
            f" | Median: {placed[feature].median():.2f}"
        )

        print(
            f"Not Placed -> Mean: {not_placed[feature].mean():.2f}"
            f" | Median: {not_placed[feature].median():.2f}"
        )

    for feature in features:
        placed_means.append(placed[feature].mean())
        not_placed_means.append(not_placed[feature].mean())

    comparison_df = pd.DataFrame({
        "Placed Mean": placed_means,
        "Not Placed Mean": not_placed_means
    }, index=features)

    comparison_df.plot(
        kind="bar",
        figsize=(14,6)
    )

    plt.title(f"{branch} - Placed vs Not Placed Feature Comparison")
    plt.xlabel("Features")
    plt.ylabel("Average Value")
    plt.xticks(rotation=45)
    plt.legend()
    plt.show()

plt.figure(figsize=(18,12))

sns.heatmap(
    df.select_dtypes(include=["int64", "float64"]).corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Feature Correlation Heatmap")
plt.show()
```
