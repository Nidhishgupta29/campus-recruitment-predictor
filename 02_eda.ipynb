```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("student_placement_prediction_dataset_2026.csv")

placed_students = df[df["placement_status"] == "Placed"]

# =========================
# Overall Placement Metrics
# =========================

total_students = len(df)
placed_count = len(placed_students)
unplaced_count = total_students - placed_count
placement_rate = (placed_count / total_students) * 100

print("===== OVERALL PLACEMENT METRICS =====")
print(f"Total Students: {total_students}")
print(f"Placed Students: {placed_count}")
print(f"Unplaced Students: {unplaced_count}")
print(f"Placement Rate: {placement_rate:.2f}%")

# ======================
# Overall Salary Metrics
# ======================

print("\n===== OVERALL SALARY METRICS =====")
print(f"Highest Package: {placed_students['salary_package_lpa'].max():.2f} LPA")
print(f"Lowest Package: {placed_students['salary_package_lpa'].min():.2f} LPA")
print(f"Average Salary: {placed_students['salary_package_lpa'].mean():.2f} LPA")
print(f"Median Salary: {placed_students['salary_package_lpa'].median():.2f} LPA")

# ==========================
# Tier-wise Placement Counts
# ==========================

print("\n===== TIER-WISE PLACEMENT COUNTS =====")
tier_placement = pd.crosstab(
    df["college_tier"],
    df["placement_status"]
)
print(tier_placement)

# ==========================
# Tier-wise Placement Rates
# ==========================

print("\n===== TIER-WISE PLACEMENT RATES (%) =====")
tier_rates = (
    df.groupby("college_tier")["placement_status"]
    .apply(lambda x: (x == "Placed").mean() * 100)
)
print(tier_rates.round(2))

# ==========================
# Tier-wise Salary Statistics
# ==========================

print("\n===== TIER-WISE SALARY STATISTICS =====")

tier_salary_stats = (
    placed_students.groupby("college_tier")["salary_package_lpa"]
    .agg(
        Highest_Package="max",
        Lowest_Package="min",
        Average_Salary="mean",
        Median_Salary="median"
    )
)

print(tier_salary_stats.round(2))

# ===========================
# Branch-wise Placement Counts
# ===========================

print("\n===== BRANCH-WISE PLACEMENT COUNTS =====")

branch_placement = pd.crosstab(
    df["branch"],
    df["placement_status"]
)

print(branch_placement)

# ===========================
# Branch-wise Placement Rates
# ===========================

print("\n===== BRANCH-WISE PLACEMENT RATES (%) =====")

branch_rates = (
    df.groupby("branch")["placement_status"]
    .apply(lambda x: (x == "Placed").mean() * 100)
)

print(branch_rates.round(2))

# ===========================
# Branch-wise Salary Statistics
# ===========================

print("\n===== BRANCH-WISE SALARY STATISTICS =====")

branch_salary_stats = (
    placed_students.groupby("branch")["salary_package_lpa"]
    .agg(
        Highest_Package="max",
        Lowest_Package="min",
        Average_Salary="mean",
        Median_Salary="median"
    )
)

print(branch_salary_stats.round(2))

# ===========================
# Salary Distribution
# ===========================

plt.figure(figsize=(8,5))
plt.hist(
    placed_students["salary_package_lpa"],
    bins=30
)

# ==========================
# Placement Distribution
# ==========================

plt.figure(figsize=(6,5))
df["placement_status"].value_counts().plot(kind="bar")
plt.title("Placement Distribution")
plt.xlabel("Placement Status")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.show()


# ==========================
# Branch Distribution
# ==========================

plt.figure(figsize=(8,5))
df["branch"].value_counts().plot(kind="bar")
plt.title("Branch Distribution")
plt.xlabel("Branch")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.show()


# ==========================
# College Tier Distribution
# ==========================

plt.figure(figsize=(6,5))
df["college_tier"].value_counts().plot(kind="bar")
plt.title("College Tier Distribution")
plt.xlabel("College Tier")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.show()


# ==========================
# Salary Distribution
# ==========================

plt.figure(figsize=(8,5))
plt.hist(
    placed_students["salary_package_lpa"],
    bins=30
)

plt.title("Salary Distribution of Placed Students")
plt.xlabel("Salary Package (LPA)")
plt.ylabel("Number of Students")
plt.show()


# ==========================
# Tier-wise Placement Counts
# ==========================

tier_placement.plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Tier-wise Placement Status")
plt.xlabel("College Tier")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.legend(title="Placement Status")
plt.show()


# ==========================
# Branch-wise Placement Counts
# ==========================

branch_placement.plot(
    kind="bar",
    figsize=(10,6)
)

plt.title("Branch-wise Placement Status")
plt.xlabel("Branch")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.legend(title="Placement Status")
plt.show()


# ==========================
# Tier-wise Average Salary
# ==========================

tier_salary_stats["Average_Salary"].plot(
    kind="bar",
    figsize=(7,5)
)

plt.title("Average Salary by College Tier")
plt.xlabel("College Tier")
plt.ylabel("Average Salary (LPA)")
plt.xticks(rotation=0)
plt.show()


# ==========================
# Branch-wise Average Salary
# ==========================

branch_salary_stats["Average_Salary"].plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Average Salary by Branch")
plt.xlabel("Branch")
plt.ylabel("Average Salary (LPA)")
plt.xticks(rotation=0)
plt.show()


# ==========================
# Tier-wise Median Salary
# ==========================

tier_salary_stats["Median_Salary"].plot(
    kind="bar",
    figsize=(7,5)
)

plt.title("Median Salary by College Tier")
plt.xlabel("College Tier")
plt.ylabel("Median Salary (LPA)")
plt.xticks(rotation=0)
plt.show()


# ==========================
# Branch-wise Median Salary
# ==========================

branch_salary_stats["Median_Salary"].plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Median Salary by Branch")
plt.xlabel("Branch")
plt.ylabel("Median Salary (LPA)")
plt.xticks(rotation=0)
plt.show()


# ==========================
# Tier-wise Highest Package
# ==========================

tier_salary_stats["Highest_Package"].plot(
    kind="bar",
    figsize=(7,5)
)

plt.title("Highest Package by College Tier")
plt.xlabel("College Tier")
plt.ylabel("Highest Package (LPA)")
plt.xticks(rotation=0)
plt.show()


# ==========================
# Branch-wise Highest Package
# ==========================

branch_salary_stats["Highest_Package"].plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Highest Package by Branch")
plt.xlabel("Branch")
plt.ylabel("Highest Package (LPA)")
plt.xticks(rotation=0)
plt.show()


# ==========================
# Tier-wise Lowest Package
# ==========================

tier_salary_stats["Lowest_Package"].plot(
    kind="bar",
    figsize=(7,5)
)

plt.title("Lowest Package by College Tier")
plt.xlabel("College Tier")
plt.ylabel("Lowest Package (LPA)")
plt.xticks(rotation=0)
plt.show()


# ==========================
# Branch-wise Lowest Package
# ==========================

branch_salary_stats["Lowest_Package"].plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Lowest Package by Branch")
plt.xlabel("Branch")
plt.ylabel("Lowest Package (LPA)")
plt.xticks(rotation=0)
plt.show()
```
